# DHA in Foods: A Comparative Analysis of the Richest Dietary Sources
## 1. Research Question
*What can food composition data reveal about the distribution of DHA across different dietary sources, and which foods stand out as the richest sources?*

### 1.1. Background: Docosahexaenoic Acid (DHA)

Docosahexaenoic acid (DHA) is a long-chain polyunsaturated fatty acid belonging to the omega-3 (n-3) family. It is an important dietary lipid that plays significant roles in the normal development and functioning of the human body, particularly in the nervous system ([Richard & Calder, 2016](https://doi.org/10.3945/an.116.012963); [Lauritzen et al., 2016](https://doi.org/10.3390/nu8010006)).

DHA is one of the three major omega-3 fatty acids commonly discussed in nutritional science, alongside alpha-linolenic acid (ALA) and eicosapentaenoic acid (EPA). While ALA is an essential fatty acid that must be obtained through the diet, the human body can produce DHA from ALA through a series of metabolic reactions, with EPA serving as an intermediate. However, this conversion is limited, making dietary intake an important source of DHA ([Richard & Calder, 2016](https://doi.org/10.3945/an.116.012963); [Lauritzen et al., 2016](https://doi.org/10.3390/nu8010006)).

From a chemical perspective, DHA is characterized by a chain of 22 carbon atoms and six cis double bonds. Its conventional notation is:

$$
C22:6n-3
$$

This notation describes three structural characteristics of the fatty acid:

- **C22** — indicates that the fatty acid contains 22 carbon atoms.
- **6** — indicates the presence of six double bonds.
- **n-3** — identifies DHA as a member of the omega-3 fatty acid family.

This chemical notation is particularly relevant to the present study because nutrient composition databases may identify DHA using this designation. Recognizing its chemical notation therefore helps establish the correct nutrient identity when locating and extracting DHA-related records from the selected databases.


## 1.2. Biological Significance of DHA

### 1.2.1. DHA During Embryonic and Fetal Development

DHA plays an important structural role during prenatal development, particularly in the development of the central nervous system and visual system. Its presence in developing neural and retinal tissues makes it an important long-chain omega-3 fatty acid during this period (Innis, 2005, [Essential Fatty Acid Transfer and Fetal Development](https://doi.org/10.1016/j.placenta.2005.01.005)).

**Placental Transfer of DHA and Maternal Supply**

During pregnancy, the developing fetus depends on the maternal supply of long-chain polyunsaturated fatty acids, including DHA. These fatty acids are transferred from the maternal circulation to the fetal circulation through the placenta. This process involves specialized fatty acid transport proteins and other mechanisms that facilitate their selective uptake and transfer (Larqué et al., 2011, [Placental Transfer of Fatty Acids and Fetal Implications](https://doi.org/10.3945/ajcn.110.001230)).

Human studies using stable isotope techniques have demonstrated preferential placental transfer of DHA compared with several other fatty acids. A higher ratio of labelled DHA in umbilical cord blood relative to maternal plasma provides evidence of this selective transfer (Larqué et al., 2011, [Placental Transfer of Fatty Acids and Fetal Implications](https://doi.org/10.3945/ajcn.110.001230)).

Maternal DHA status, including dietary intake and existing body stores, contributes to the amount of DHA available for placental transfer. Therefore, maternal nutritional status is an important factor influencing fetal DHA supply (Larqué et al., 2011, [Placental Transfer of Fatty Acids and Fetal Implications](https://doi.org/10.3945/ajcn.110.001230)).

**DHA Accumulation in the Developing Fetal Brain**

DHA is a major structural component of neuronal membranes and contributes to the development and functional organization of the fetal brain. Its incorporation into membrane phospholipids supports the structural properties of developing neural cells and is relevant to neuronal membrane function (Lauritzen et al., 2016, [DHA Effects in Brain Development and Function – Section 2](https://pmc.ncbi.nlm.nih.gov/articles/PMC4728620/)).

The accumulation of DHA in the fetal brain increases substantially during the third trimester of pregnancy. This period coincides with rapid brain growth and the development of important neuronal structures. The availability of maternal long-chain omega-3 fatty acids and their transfer across the placenta are therefore important for fetal DHA accumulation (Lauritzen et al., 2016, [Section 2: Brain DHA Accumulation during Development Depending on Diet](https://pmc.ncbi.nlm.nih.gov/articles/PMC4728620/)).

**DHA and Prenatal Visual System Development**

In addition to its role in the developing brain, DHA is an important structural component of retinal tissue, particularly the membranes of photoreceptor cells. These specialized cells detect light and initiate visual signaling. The presence of DHA in retinal membranes contributes to their structural and functional properties (Innis, 2005, [Essential Fatty Acid Transfer and Fetal Development](https://doi.org/10.1016/j.placenta.2005.01.005)).


### 1.2.2. DHA During Lactation and Early Infancy

Following birth, DHA continues to play an important role in early development. Human breast milk naturally contains DHA, providing a dietary source of this long-chain omega-3 fatty acid during infancy.

The DHA concentration in human milk is influenced by maternal nutritional status and dietary intake. A recent systematic review identified a positive association between maternal DHA consumption and DHA concentrations in breast milk, indicating that the mother's diet can influence the amount of DHA available to the nursing infant (Petersohn et al., 2024, [Maternal Diet and Human Milk Composition: An Updated Systematic Review](https://doi.org/10.3389/fnut.2023.1320560)).

Breast milk is an important source of DHA during early infancy and contributes to its accumulation in developing tissues. Studies reviewed by Lauritzen et al. (2016) have reported higher DHA levels in the brain cortex and red blood cells of breast-fed infants compared with age-matched formula-fed infants, highlighting the contribution of breast milk to postnatal DHA supply (Lauritzen et al., 2016, [Section 2: Brain DHA Accumulation during Development Depending on Diet](https://pmc.ncbi.nlm.nih.gov/articles/PMC4728620/)).

The nutritional significance of DHA in early infancy is also reflected in European legislation. Following the scientific recommendations of the European Food Safety Authority (EFSA), Commission Delegated Regulation (EU) 2016/127, adopted on 25 September 2015 and applicable from 22 February 2020, introduced the mandatory inclusion of DHA in both infant formula and follow-on formula, establishing a required range of 20–50 mg per 100 kcal ([EFSA NDA Panel, 2014](https://doi.org/10.2903/j.efsa.2014.3760); [European Commission, 2016](https://eur-lex.europa.eu/eli/reg_del/2016/127/oj)).

### 1.2.3. DHA in Neurodevelopment and Cellular Function

DHA is a major structural component of membrane phospholipids in the central nervous system, where it contributes to membrane organization and influences membrane-dependent cellular functions. In addition to its structural role, DHA serves as a precursor of bioactive lipid mediators and participates in cellular signaling pathways. Experimental evidence from cell and animal studies supports its involvement in neuronal growth, differentiation, and signaling processes ([Richard & Calder, 2016](https://doi.org/10.3945/an.116.012963); [Lauritzen et al., 2016](https://doi.org/10.3390/nu8010006)).

Although the rate of DHA accumulation in the brain decreases after early development, its incorporation continues during childhood, particularly throughout the preschool years. High DHA levels in the brain are generally maintained throughout later life, although this may depend on an adequate dietary supply. However, evidence concerning the effects of DHA intake on cognitive performance in older children remains limited and inconsistent, and current findings do not establish a general cognitive benefit of increased DHA intake in this age group ([Lauritzen et al., 2016](https://doi.org/10.3390/nu8010006)).

## 2. Research Objectives and Scope

### 2.1. Research Aim

The aim of this study is to investigate and compare the docosahexaenoic acid (DHA) content of foods using two national food composition databases: **Matvaretabellen (Norway)** and **FOODfiles 2024 (New Zealand)**. The study seeks to identify foods with the highest recorded DHA concentrations, examine the distribution of DHA across different food categories, and explore similarities and differences between the two databases.

The two databases are selected because they are established national food composition resources that provide extensive, structured information on the nutritional composition of foods. Matvaretabellen provides data on more than 2,000 foods, while FOODfiles 2024 includes 2,857 food records and offers an extensive range of nutritional components in its Unabridged version. Their complementary data structures and different national contexts make them suitable for a comparative investigation of DHA concentrations across a broad range of foods. In addition, their documented data sources and nutrient information allow the origin and characteristics of the reported values to be examined as part of the analysis.

### 2.2. Research Objectives

To achieve the research aim, the study pursues the following objectives:

1. Retrieve and extract DHA-related records from both food composition databases.
2. Examine the structure, variables, units of measurement, and provenance of the available data.
3. Assess data quality by identifying missing values, zero values, potential inconsistencies, and differences in data reporting.
4. Clean, transform, and standardize the datasets to support reliable analysis.
5. Explore the distribution of DHA concentrations across different food categories using descriptive statistics and data visualizations.
6. Identify the ten foods with the highest recorded DHA concentrations in each database.
7. Compare the findings from the two databases, taking into account differences in food descriptions, data sources, and measurement characteristics.
8. Interpret the results and discuss the limitations affecting the comparability and reliability of the findings.

### 2.3. Scope of the Study

The analysis focuses on foods recorded in Matvaretabellen and FOODfiles 2024 that contain DHA-related composition data.

The scope is defined by the following criteria:

- **Food categories:** All categories of foods are considered, including fish and seafood, meat, eggs, dairy products, and other food groups, provided that relevant DHA data are available.
- **Exclusions:** Dietary supplements and concentrated products intended primarily for supplementation are excluded from the main analysis.
- **Nutrient focus:** The study examines DHA specifically, rather than total omega-3 fatty acids or other individual fatty acids.
- **Measurement basis:** DHA concentrations will be assessed using a consistent measurement basis, subject to verification of the units and reporting conventions documented in each database.
- **Comparative approach:** Each database will be analysed separately before the results are compared. The study will not combine the two databases into a single ranking.
- **Interpretation:** The analysis concerns DHA concentrations reported in food composition databases. It does not estimate actual dietary intake or establish health effects associated with DHA consumption.
